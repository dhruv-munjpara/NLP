```python

from gensim.models import FastText

# Sample sentences
sentences = [
    ["playing", "football", "outside"],
    ["playing", "cricket", "outside"],
    ["player", "plays", "football"],
    ["football", "is", "fun"]
]

# Train FastText model
model = FastText(
    sentences,
    vector_size=5,
    window=2,
    min_count=1,
    workers=1,
    seed=42,
    epochs=100
)

# Get vector of a word
print("Vector of playing:")
print(model.wv["playing"])

# Find similar words
print("\nWords similar to playing:")
print(model.wv.most_similar("playing", topn=3))

# Test a new word
print("\nVector of unseen word:")
print(model.wv["playingly"])

#playing 

#multiple pattern 
```
