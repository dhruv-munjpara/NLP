```python

from gensim.models import Word2Vec
#semantic search // google LLM -gemini //NLP
#feature extraction = word2vec 
#meta=fasttext 
# Sample sentences
sentences = [
    ["i", "love", "python"],
    ["i", "love", "machine", "learning"],
    ["python", "is", "easy"],
    ["machine", "learning", "is", "interesting"],
    ["i", "love", "learning"]
]
# = no of 5 python=[0.23,0.78,90,.34,.35]
# Train Word2Vec model
#cbow target  skip gram this  xyz
# wor single time consider 
model = Word2Vec(
    sentences,
    vector_size=5,
    window=2,
    min_count=1,
    workers=1, #cpu -multiple process =port 1aconsume gpu 
    seed=42,
    epochs=100
)
#live application = customer data 
#seed force add chatboat =order  3
# Get vector of a word
#random state =42 
print("Vector of Python:")
print(model.wv["python"])

# Find similar words
print("\nWords similar to Python:")
print(model.wv.most_similar("python", topn=3))
```
