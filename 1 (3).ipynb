```python
import re
import nltk
```

```python

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from sklearn.feature_extraction.text import CountVectorizer
```

```python
# Download required resources
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
text = "I am loving Natural Language Processing! I am learning NLP."
print("Original Text:")
print(text)
```

```python
# 2. Text Cleaning
clean_text = text.lower()
clean_text = re.sub(r'[^a-zA-Z\s]', '', clean_text)

# regular expression //pattern undestand data convert 
print("\nCleaned Text:")
print(clean_text)
```

```python

# 3. Tokenization
tokens = word_tokenize(clean_text)

print("\nTokens:")
print(tokens)
```

```python

# 4. Stopword Removal
stop_words = set(stopwords.words('english'))

filtered_words = [
    word for word in tokens
    if word not in stop_words
]

print("\nAfter Stopword Removal:")
print(filtered_words)
```

```python

# 5. Stemming root 
stemmer = PorterStemmer()

stemmed_words = [
    stemmer.stem(word)
    for word in filtered_words
]

print("\nAfter Stemming:")
print(stemmed_words)
```

```python

# 6. Lemmatization
lemmatizer = WordNetLemmatizer()

lemmatized_words = [
    lemmatizer.lemmatize(word)
    for word in filtered_words
]

print("\nAfter Lemmatization:")
print(lemmatized_words)
```

```python
#7. Feature Extraction (Bag of Words)
print("\nFeature Extraction - Bag of Words")
#bag of count 2 1 3
# Convert processed words into a sentence
processed_text = " ".join(lemmatized_words)

# CountVectorizer requires text documents
vectorizer = CountVectorizer()

X = vectorizer.fit_transform([processed_text])

# Display features
print("\nFeature Names:")
print(vectorizer.get_feature_names_out())

# Display feature matrix
print("\nFeature Matrix:")
print(X.toarray())
```
